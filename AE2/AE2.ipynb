# r = 1200

```python
import copy
rectangles1 = pd.read_csv("cutting/r1200.csv", names=["width", "height", "value"])
rectangles1['width'] = rectangles1['width']/1200
rectangles1['height'] = rectangles1['height']/1200
rectangles2 = copy.deepcopy(rectangles1)
rectangles1np = rectangles1.values
rectangles2 = rectangles2[['height', 'width', 'value']]
rectangles2np = rectangles2.values
rectangles = np.vstack((rectangles1np, rectangles2np))


# Create the packer instance
packer = CircleRectanglePacker(
    rectangle_types=rectangles,
    population_size=300,
    initial_mutation_rate=0.5,
    initial_crossover_rate=0.9,
    elite_size=15,
    initial_tournament_size=5,
    max_generations=300
)

# Run the algorithm
best_solution, best_fitness = packer.evolve()

# Visualize the result
print(f"\nBest solution found with total value: {best_fitness}")
print(f"Solution contains {len(best_solution)} rectangles")
print(f"Fill ratio: {packer.calculate_filled_ratio(best_solution):.4f}")

packer.visualize_solution(best_solution)
```
